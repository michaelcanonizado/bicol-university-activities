```python
import numpy as np
import matplotlib.pyplot as plt

def show_color(r, g, b):
    color = np.zeros((100, 100, 3), dtype=np.uint8)
    color[:, :] = [r, g, b]
    plt.imshow(color)
    plt.axis("off")
    plt.title(f"RGB = ({r}, {g}, {b})")
    plt.show()

show_color(255, 100, 50)
```

```python
from google.colab import files
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

print("Upload 2 images with different lighting")

# Upload 2 images
uploaded = files.upload()

filenames = list(uploaded.keys())

if len(filenames) != 2:
    raise ValueError("Please upload exactly 2 images.")

# Open both images and convert to RGB
image1 = Image.open(filenames[0]).convert("RGB")
image2 = Image.open(filenames[1]).convert("RGB")

# Make sure both images have the same dimensions
if image1.size != image2.size:
    raise ValueError("The two images must have the same dimensions.")

# Convert to NumPy arrays
image1_array = np.array(image1)
image2_array = np.array(image2)

print("Image 1 shape:", image1_array.shape)
print("Image 2 shape:", image2_array.shape)

# Display the two images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image1)
plt.title("Image 1")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(image2)
plt.title("Image 2")
plt.axis("off")

plt.show()


# Get image dimensions
height, width, _ = image1_array.shape

# Choose 5 random locations
np.random.seed(42)  # Makes the same 5 locations repeatable

print("\nRGB comparison at the same locations:")
print("-" * 50)

for i in range(5):

    # Random row and column
    row = np.random.randint(0, height)
    col = np.random.randint(0, width)

    # Get RGB values from both images at the SAME location
    rgb1 = image1_array[row, col]
    rgb2 = image2_array[row, col]

    print(f"\nPixel {i + 1}")
    print(f"Location: row={row}, column={col}")
    print(f"Image 1 RGB: {rgb1}")
    print(f"Image 2 RGB: {rgb2}")
```
